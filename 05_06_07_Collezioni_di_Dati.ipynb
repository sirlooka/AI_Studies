### INDEXING

```python
### **INDEXING**
l = [1, 2, 3, 4, 5, 6]
l[0] # quindi nome_lista[index_da_0] per ottenere l'elemento in quell'index
l[-3] # per partire dalla fine della lista \\ l'indexing parte da 1 in questo caso
```

### **SLICING** - sezionare le liste

Per ottenere più valori da una lista si usa lo **slicing**

```python
l = [1, 2, 3, 4, 5, 6]
l[1:3] #si specifica l'intervallo di indici da considerare ESCLUSO L'ULTIMO
# risultato [2, 3]
```

```python

l[0:5] #si può scrivere anche l[:5]
# risultato [1, 2, 3, 4, 5]
```

```python
l[:] # viene selezionato TUTTO il contenuto
```

```python
l[-2:] # dagli ultimi due fino alla fine
```

```python
l[::-1] # accedo agli elementi dall'ultimo al primo invertendo il contenuto della lista
```

### APPEND - aggiungere oggetti alla lista

### **POP** - estrazione oggetti
La funzione **POP** rimuove un elemento dalla lista e lo ESTRAE per ritornarlo come valore



```python
l = [1, 2, 3, 4, 5, 6]
var = l.pop(0) #estrae il valore con indice 0 e lo inserisce nella var
print(var, l) #stampa var e la lista completa
```

se vogliamo vedere all'interno di una lista se c'è qualche oggetto, possiamo cercarlo con la funzione IN

```python
subjects = ["Machine Learning", "Math", "Cicciopasticcio", "Pippo Baudo"]
"Machine Learning" in subjects

# risultato= True perché compare nella lista - case sensitive ricorda
# cosa succede se ce ne fossero due?

subjects = ["Machine learning", "Math", "Cicciopasticcio", "Pippo Baudo"]
subjects.append ("Machine learning")
"Machine learning" in subjects
# risultato = True perché compare nella lista - sempre True ovviamente

subjects.index("Machine learning")
# ris. 0 -> prima posizione con indexing
```

Le liste annidate sono liste che contengono altre liste al loro interno. Sono utili per rappresentare dati strutturati, come matrici o tabelle.

Possiamo anche iterare su liste annidate usando cicli `for`.

# LISTE ANNIDATE

Le liste annidate si comportano in modo simile alle liste normali per quanto riguarda l'accesso agli elementi e l'iterazione, ma richiedono l'uso di indici multipli per accedere agli elementi interni alle liste contenute. Ad esempio, nested_list[1][1] accede all'elemento nella seconda lista (indice 1) e poi al secondo elemento all'interno di quella lista (indice 1).

```python
# Esempio di lista annidata
nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Per accedere agli elementi in una lista annidata, si usano indici multipli
# Ad esempio, per accedere al 5 (che si trova nella seconda lista, secondo elemento):
element = nested_list[1][1]
print(element)

# Per accedere alla prima lista:
first_inner_list = nested_list[0]
print(first_inner_list)
```

```python
# Iterare su una lista annidata
for inner_list in nested_list:
  for element in inner_list:
    print(element)
```
